```python
#Name: Douglas Marshall
#ASS:Mod 3 Tutorial
#DEF: Looks for a number in the elements of an array and returns that number






class Solution:
    def binarysearch(self, arr, k):
        low = 0
        high = len(arr) - 1
        
        while low <= high:#set mid to middle
    
            mid = low + (high - low) // 2
    
            # Check if mid is in right spot
            if arr[mid] == k:
                return mid
    
            # If k is greater, ingnore
            elif arr[mid] < k:
                low = mid + 1
    
            # If x is smaller, ignore right half
            else:
                high = mid - 1
    
        # If we reach here, then the element
        # was not present
        return -1
```
