Description: Guess a number between 1 and 10 til you get it right

```python
import random


secret = random.randint(1, 10)

guess = int(input('Please guess a number between 1 and 10: '))

while guess != secret:
    if guess > 10 or guess < 1:
        guess = int(input('Invalid input. Please guess a number between 1 and 10: '))
        
    elif guess > secret:
        guess = int(input('too high'))
    
    elif guess < secret:
        guess = int(input('too low'))
        
print('just right')
```
