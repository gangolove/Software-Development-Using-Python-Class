Description: Guess a number between 1 and 10 til you get it right

```python
guess_me = 7
number = 1

while True: 
    if number < guess_me:
        print('too low')
    
    if number == guess_me:
        print('found it!')
        break
        
    if number > guess_me:
        print('oops')
        break
    
    number += 1
```
